```python
import requests
base_url = "http://localhost:8002"
```

```python
update_path = "/api/events/12/"
update_endpoint = f"{base_url}{update_path}"
headers = {
    "Content-Type": "application/json"
}
response = requests.put(update_endpoint, headers=headers, json={"page": "/test-", "description": "hello world"})
print(response.headers)
if response.ok:
    data = response.json()
    print(data, type(data), data.get("id"))
else: 
    print(response.text)
    print(False)
```

```python
create_path = "/api/events/"
create_endpoint = f"{base_url}{create_path}"
data = {"page": "/test+", "description": "hello user"}
response = requests.post(create_endpoint, headers=headers, json=data)
print(response.headers)
if response.ok:
    data = response.json()
    print(data, type(data), data.get("id"))
else: 
    print(response.text)
    print(False)
```
