```python
import requests
from pprint import pprint
path = "/api/events/"
base_url = "http://localhost:8002"
endpoint = f"{base_url}{path}"
endpoint
```

```python
response = requests.get(endpoint, params={'duration': '1 minute', 'pages': []})
if response.ok:
    data = response.json()
    pprint(data)
else: 
    print(False)
```

```python
detail_path = "/api/events/1"
detail_endpoint = f"{base_url}{detail_path}"
response = requests.get(detail_endpoint)
if response.ok:
    data = response.json()
    pprint(data)
else: 
    print(response.status_code, response.text)
```
